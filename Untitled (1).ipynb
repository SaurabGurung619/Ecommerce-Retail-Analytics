# Convert dates

Currently dates are stored as strings/objects. ML and time-series analysis require proper datetime values.

```python
customers["signup_date"] = pd.to_datetime(
    customers["signup_date"],
    errors="coerce"
)

orders["order_date"] = pd.to_datetime(
    orders["order_date"],
    errors="coerce"
)

orders["delivery_date"] = pd.to_datetime(
    orders["delivery_date"],
    errors="coerce"
)

orders["estimated_delivery_date"] = pd.to_datetime(
    orders["estimated_delivery_date"],
    errors="coerce"
)
```

# Join Orders + Products

## Why?
Our prediction is about product demand, so we need the product characteristics together with order history.

```python
sales = orders.merge(
    products,
    on="product_id",
    how="left",
    suffixes=("_order", "_product")
)
sales.shape
```

```python
sales.head()
```

# Filter valid orders
## Why?
Our target is actual fulfilled product demand.
- A cancelled order should not be counted as successfully sold demand.

```python
orders["order_status"].value_counts()
```

```python
sales = sales[
    sales["order_status"] == "Delivered"
].copy()
```

```python
(sales["order_status"] == "Delivered").value_counts()
```

## Aggregate demand by Product + Month
This is the most important transformation.
### Why?
The original Orders table is transaction-level:

But our question is:
How much demand did Product A have in each month?
So we convert:
Order-level data
       ↓
Product + Month
       ↓
Monthly demand

```python
sales["month"] = (
    sales["order_date"]
    .dt.to_period("M")
    .dt.to_timestamp()
)
```

```python
monthly_demand = (
    sales
    .groupby(["product_id", "month"])
    .agg(
        demand=("quantity", "sum"),
        revenue=("total_amount", "sum"),
        avg_price=("unit_price_order", "mean"),
        avg_discount=("discount_pct", "mean"),
        order_count=("order_id", "nunique")
    )
    .reset_index()
)
```

```python
monthly_demand
```

## Perform demand analysis
### Why?
Before ML, we need to understand the target variable.

```python
# Total demand
monthly_demand["demand"].sum()
```

```python
# Average monthly demand
monthly_demand["demand"].mean()
```

# Demand by category
First join product information:

```python
product_features = products[
    [
        "product_id",
        "category",
        "sub_category",
        "brand",
        "rating"
    ]
].drop_duplicates("product_id")

monthly_demand = monthly_demand.merge(
    product_features,
    on="product_id",
    how="left"
)
```

```python
category_demand = (
    monthly_demand
    .groupby("category")["demand"]
    .sum()
    .sort_values(ascending=False)
)

category_demand
```

# Create ML features
### Why?
The Random Forest needs input variables/features that can help predict future demand.

We need information such as:
- Previous demand
- 3-month demand history
- 6-month demand history
- Price
- Discount
- Number of orders
- Rating
- Category
- Brand
- Month
- Quarter

```python
months = pd.date_range(
    orders["order_date"].min().to_period("M").to_timestamp(),
    orders["order_date"].max().to_period("M").to_timestamp(),
    freq="MS"
)

grid = pd.MultiIndex.from_product(
    [products["product_id"].unique(), months],
    names=["product_id", "month"]
).to_frame(index=False)

panel = grid.merge(
    monthly_demand,
    on=["product_id", "month"],
    how="left"
)
```

```python
panel["demand"] = panel["demand"].fillna(0)

panel["revenue"] = panel["revenue"].fillna(0)

panel["order_count"] = panel["order_count"].fillna(0)

panel["avg_discount"] = panel["avg_discount"].fillna(0)

panel["avg_price"] = panel["avg_price"].fillna(
    panel["avg_price"].median()
)

panel["rating"] = panel["rating"].fillna(
    panel["rating"].median()
)

panel["category"] = panel["category"].fillna("Unknown")

panel["sub_category"] = panel["sub_category"].fillna("Unknown")

panel["brand"] = panel["brand"].fillna("Unknown")
```

```python
panel
```

This is a much better structure for demand prediction.

# Create historical demand features
Why?
The strongest clue about future demand is often past demand.

```python
panel = panel.sort_values(
    ["product_id", "month"]
)
```

```python
panel["previous_month_demand"] = (
    panel
    .groupby("product_id")["demand"]
    .shift(1)
)
```

```python
panel["previous_3_month_avg"] = (
    panel
    .groupby("product_id")["demand"]
    .transform(
        lambda x:
        x.shift(1).rolling(3).mean()
    )
)
```

```python
panel["previous_6_month_avg"] = (
    panel
    .groupby("product_id")["demand"]
    .transform(
        lambda x:
        x.shift(1).rolling(6).mean()
    )
)
```

## Why shift(1)?
This is extremely important.
## We want:
Past
 ↓
Model
 ↓
Future

# Create next-month target
This is where we tell Random Forest what it needs to predict.

```python
panel["target_demand"] = (
    panel
    .groupby("product_id")["demand"]
    .shift(-1)
)
panel
```

# Remove data leakage
## Why?
We must make sure the model doesn't see future information.

```python
ml_data = panel.dropna(
    subset=[
        "previous_month_demand",
        "previous_3_month_avg",
        "previous_6_month_avg",
        "target_demand"
    ]
).copy()
ml_data
```

# Train/Test split by time

# Why?
This is a future prediction problem.
- We should never randomly mix 2025 data into training and testing.
- We'll use:-
TRAIN
- 2023-07 → 2025-09

TEST
- 2025-10 → 2025-11

```python
cutoff_date = pd.Timestamp("2025-10-01")

train = ml_data[
    ml_data["month"] < cutoff_date
].copy()

test = ml_data[
    ml_data["month"] >= cutoff_date
].copy()
```

```python
print(train.shape)
print(test.shape)
```

# Encode categorical features
Random Forest cannot directly understand:
- category = Electronics
- brand = Nova
- sub_category = Laptop

So we convert them into numerical representation.

```python
features = [
    "previous_month_demand",
    "previous_3_month_avg",
    "previous_6_month_avg",
    "avg_price",
    "avg_discount",
    "order_count",
    "rating",
    "month_number",
    "quarter",
    "category",
    "sub_category",
    "brand"
]
```

```python
train["month_number"] = train["month"].dt.month
train["quarter"] = train["month"].dt.quarter

test["month_number"] = test["month"].dt.month
test["quarter"] = test["month"].dt.quarter
```

```python
missing = [col for col in features if col not in train.columns]

print("Missing columns:", missing)
```

```python
panel["month_number"] = panel["month"].dt.month
panel["quarter"] = panel["month"].dt.quarter
```

```python
categorical_features = [
    "category",
    "sub_category",
    "brand"
]

numeric_features = [
    "previous_month_demand",
    "previous_3_month_avg",
    "previous_6_month_avg",
    "avg_price",
    "avg_discount",
    "order_count",
    "rating",
    "month_number",
    "quarter"
]
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer

preprocessor = ColumnTransformer(
    transformers=[
        ("num", SimpleImputer(strategy="median"), numeric_features),
        ("cat", OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=False
        ), categorical_features)
    ]
)
```

# Train Random Forest
## Why Random Forest?
Random Forest combines many decision trees.

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.pipeline import Pipeline

X_train = train[features]
y_train = train["target_demand"]

X_test = test[features]
y_test = test["target_demand"]
```

```python
rf = RandomForestRegressor(
    n_estimators=100,
    max_depth=20,
    min_samples_leaf=2,
    random_state=42,
    n_jobs=-1
)
```

```python
X_train.isna().sum()
```

```python
print("Rows in train:", len(train))

print("\nMissing values:")
print(train[[
    "avg_price",
    "rating",
    "category",
    "sub_category",
    "brand"
]].isna().sum())
```

# Create pipeline:

```python
rf = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", rf)
    ]
)
```

```python
model.fit(
    X_train,
    y_train
)
```

```python
y_pred = model.predict(X_test)

print("Prediction completed!")
```

```python
results = test[
    ["product_id", "month"]
].copy()

results["actual_demand"] = y_test.values
results["predicted_demand"] = y_pred

results.head(10)
```

# Evaluate the model

```python
from sklearn.metrics import mean_absolute_error

mae = mean_absolute_error(
    y_test,
    y_pred
)

print("MAE:", mae)
```

```python
from sklearn.metrics import mean_squared_error

rmse = np.sqrt(
    mean_squared_error(y_test, y_pred)
)

print("RMSE:", rmse)
```

```python
from sklearn.metrics import r2_score

r2 = r2_score(
    y_test,
    y_pred
)

print("R²:", r2)
```

```python
# ============================================
# PRODUCT STOCK vs PREDICTED DEMAND
# ============================================

# 1. Create prediction results
results = test[[
    "product_id",
    "month"
]].copy()

results["actual_demand"] = y_test.values
results["predicted_demand"] = y_pred


# 2. Add current stock from Products table
results = results.merge(
    products[
        [
            "product_id",
            "stock_quantity"
        ]
    ],
    on="product_id",
    how="left"
)


# 3. Calculate stock gap
results["stock_gap"] = (
    results["stock_quantity"]
    - results["predicted_demand"]
)


# 4. Calculate additional stock needed
results["additional_stock_needed"] = (
    results["predicted_demand"]
    - results["stock_quantity"]
).clip(lower=0)


# 5. Create business recommendation
results["stock_status"] = np.where(
    results["stock_gap"] < 0,
    "Need More Stock",
    "Stock Sufficient"
)


# 6. Round prediction values
results["predicted_demand"] = (
    results["predicted_demand"]
    .round()
    .astype(int)
)

results["additional_stock_needed"] = (
    results["additional_stock_needed"]
    .round()
    .astype(int)
)


# 7. Display final result
final_stock_report = results[
    [
        "product_id",
        "month",
        "actual_demand",
        "stock_quantity",
        "predicted_demand",
        "additional_stock_needed",
        "stock_status"
    ]
]

print(final_stock_report.head(20))
```

10. Create ML Features
We create features that help the Random Forest understand patterns in product demand. These include previous-month demand, previous 3-month and 6-month average demand, average price, discount, order count, product rating, month, quarter, category, sub-category, and brand. These features represent the historical information available before the month we want to predict.
11. Create Next-Month Target
We create target_demand by shifting each product's demand one month forward. This means the model uses information from previous months to predict the following month's demand. For example, January's historical information is used to predict February demand, February information predicts March demand, and so on.
12. Remove Data Leakage
We make sure the model does not receive information from the future. Features must contain only information that would have been available when the prediction was made. We therefore use shift(1) for historical demand features and remove rows where the required historical information or next-month target is unavailable. This makes the model evaluation realistic.
13. Train/Test Split by Time
Because this is a future-demand prediction problem, we split the data chronologically instead of randomly. Earlier months are used for training, while later months are kept completely unseen for testing. For example, we can train using historical data up to September 2025 and evaluate the model on October and November 2025.
14. Encode Categorical Features
Random Forest requires numerical inputs, so categorical variables such as category, sub_category, and brand are converted into numerical representations using One-Hot Encoding. Numerical features such as previous demand, price, discount, rating, and order count are passed directly to the model.
15. Train Random Forest
We train a RandomForestRegressor using the prepared training data. Random Forest creates many decision trees and combines their predictions to produce the final demand prediction. It is useful here because product demand may have nonlinear relationships with factors such as historical demand, price, discount, category, and seasonality.
16. Predict
After training, we provide the model with the unseen test data. The Random Forest produces a predicted demand value for every product-month in the test period. For example, if a product currently has 500 units of stock, the model might predict that demand for the next month will be 600 units.
17. Evaluate the Model
We compare the predicted demand with the actual demand using MAE, RMSE, and R². MAE tells us the average number of units by which our predictions are wrong, RMSE gives more weight to large errors, and R² tells us how much variation in demand the model explains. These metrics help determine whether the Random Forest is actually useful rather than assuming it is good simply because it is a powerful algorithm.
18. Tune Random Forest
After establishing a baseline model, we tune parameters such as n_estimators, max_depth, min_samples_split, min_samples_leaf, and max_features. The objective is to improve prediction performance without overfitting. Because this is time-based data, the validation strategy should also respect chronological order.
19. Compare Actual vs Predicted
We create a table and visualization comparing actual demand with predicted demand. This helps us understand where the model performs well and where it makes larger errors. For example, we can compare a product's actual demand of 600 units with the model's prediction of 570 units.
20. Feature Importance
Random Forest allows us to examine feature importance and identify which variables contribute most to the predictions. We can determine whether historical demand, price, discount, product rating, seasonality, category, or other variables are important for predicting demand. This provides useful business insights in addition to the prediction itself.
21. Business Recommendations
Finally, we compare predicted demand with current stock quantity from the Products table. If current stock is 500 units and predicted demand is 600 units, there is a potential shortage of 100 units, so the business could consider obtaining additional inventory. If stock is 800 and predicted demand is 500, the existing stock may be sufficient. This converts the ML prediction into an actionable inventory recommendation.

Interview answer
